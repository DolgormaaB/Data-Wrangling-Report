##### **Task 9:** Let's deal with duplicate data

````python
airbnb.groupby('room_type')['price'].max()
````


There are two notable types of duplicate data:

- Identical duplicate data across all columns
- Identical duplicate data cross most or some columns

To diagnose, and deal with duplicate data, we will be using the following methods and functions:

- `.duplicated(subset = , keep = )`
  - `subset` lets us pick one or more columns with duplicate values.
  - `keep` returns lets us return all instances of duplicate values.
- `.drop_duplicates(subset = , keep = )`
  

````python
# Print the header of the DataFrame again
airbnb.head()
````

````python
# Find duplicates
dup = airbnb[airbnb.duplicated()]
dup.head(5)
````

````python
# Remove identical duplicates
airbnb = airbnb.drop_duplicates(keep='first')
airbnb.duplicated().sum()
````

````python
# Find non-identical duplicates
non_identical_dup = airbnb[airbnb.duplicated(subset=['latitude', 'longitude', 'host_name'], keep=False)]
non_identical_dup.head()
````

````python
# Show all duplicates
non_identical_dup.sort_values(by = "host_name")
````

To treat identical duplicates across some columns, we will chain the `.groupby()` and `.agg()` methods where we group by the column used to find duplicates (`listing_id`) and aggregate across statistical measures for `price`, `rating` and `list_added`. The `.agg()` method takes in a dictionary with each column's aggregation method - we will use the following aggregations:

- `mean` for `price` and `rating` columns
- `max` for `listing_added` column
- `first` for all remaining column

*A note on dictionary comprehensions:*

Dictionaries are useful data structures in Python with the following format
`my_dictionary = {key: value}` where a `key` is mapped to a `value` and whose `value` can be returned with `my_dictionary[key]` - dictionary comprehensions allow us to programmatically create dicitonaries using the structure:

```
{x: x*2 for x in [1,2,3,4,5]}
{1:2, 2:4, 3:6, 4:8, 5:10}
```

````python
# create a dictionary of aggregations for all columns except listing_id
aggregations = {col: 'first' for col in airbnb.columns if col != 'listing_id'} # all 'first' at first
aggregations['price'] = 'mean' # for price we want to take mean
aggregations['rating'] = 'mean' # for rating we want to take mean
aggregations['listing_added'] = 'max' # for listing_added we want to take the maximum date
airbnb = airbnb.groupby('listing_id').agg(aggregations).reset_index() # group by listing_id and apply appropriate aggregations for all columns
airbnb.head()
````

## Categorical Variables
For nominal data like room_type or borough we cant calculate mean or standard deviation, so we instead look at frequencies, percentages, and mode.

## Continuous Variables

When it comes to continuous numerical data we can use all our mathematical oeprations like Central Tendency i.e mean and median, Variabiliy i.e Standarad Deviation and IQR, and Shape i.e skweness and kurtosis.

````python
num_vars = ['price', 'rating', 'availability_365']

for var in num_vars:
    print(f"{var.upper()}")
    #central tendency
    print(f"Mean:   {airbnb[var].mean():.2f}")
    print(f"Median: {airbnb[var].median():.2f}")

    # variability
    print(f"Std Dev:{airbnb[var].std():.2f}")
    q1 = airbnb[var].quantile(0.25)
    q3 = airbnb[var].quantile(0.75)
    print(f"IQR:    {q3 - q1:.2f}")

    #shape
    print(f"Skew:   {airbnb[var].skew():.2f}")
    print(f"Kurt:   {airbnb[var].kurt():.2f}\n")
````

````python
fig, axes = plt.subplots(3, 2, figsize=(14, 15))

for i, var in enumerate(num_vars):
    # histogram
    sns.histplot(airbnb[var].dropna(), kde=True, ax=axes[i, 0], color='skyblue')
    axes[i, 0].axvline(airbnb[var].mean(), color='red', linestyle='--', label='Mean')
    axes[i, 0].axvline(airbnb[var].median(), color='green', linestyle='-', label='Median')
    axes[i, 0].set_title(f'{var.capitalize()} Distribution')
    axes[i, 0].legend()

    # boxplot
    sns.boxplot(x=airbnb[var].dropna(), ax=axes[i, 1], color='lightgreen')
    axes[i, 1].set_title(f'{var.capitalize()} Boxplot (Spread & Outliers)')

plt.tight_layout()
plt.show()
````

###Interpretation:

#### 1. Price

i.) Central Tendency : the mean is much higher than the median indicating the data is heavily skewed by expensive outliers.

ii.) Variabliliy: the standard deviation is large i.e roughly $203 showing massive dispersion.

iii.) Shape : There is extremely high positive skewness (14.6) and kurtosis (373.7). It has a very long right tail (prices up to $8,000) making the median a much better measure of typical price than the mean.
The high kurtosis value tells us that a majority of hosts are setting their rental prices at around the same rate of 100 to 150 (due to how 'spiky' our data is in the middle).

#### 2. Rating :

i.) : The mean and median are almost identical, i.e about 4.

ii.) : The standard deviation stands at 0.57 which implies that most of the listings deviate from the 4.01 average. The IQR of 1 indicates that the middle 50% of all the airbnbs in NY are separated by just one star (mostly between 4 and 5)

iii.) : the distribution is relatively flat with a slight negative skew (-0.04) which means the scores are fairly spread otu but cluster a bit more toward the higher ratings. The negative kurtosis (-1.19) means the typical 'bell curve' peak isn't present here and has lighter tails (no extreme outliers past 3 to 5 range)

#### 3. Availability :

i.) : The mean is 112 days, however the median is only 44 days. The mode is 0, meaning a huge chunk of properties are barely even available.

ii.) : The IQR here is 226 days, meaning that the availability varies widly between hosts.

iii.) : the skewness is positive (0.77), implying a heavy clustering near zero days avaialable. With a smaller peak at 365 days (the hosts that rent year round.)


## Cross Sectional Analysis
Just like in Exercise 8, we're gonna group our data by a category and calculate things like **skewness** and **kurtosis**.

Let's group the Airbnb **prices** by **borough** to see how the distributions look across different areas of New York.

````python
grouped_price = airbnb.groupby('borough')['price']

grouped_summary = grouped_price.describe()

# skewness and kurtosis
grouped_summary['Skewness'] = grouped_price.apply(lambda x: x.skew())
grouped_summary['Kurtosis'] = grouped_price.apply(lambda x: x.kurt())

grouped_summary
````

## Room Type Price (Median, IQR)

````python

grouped_price_rt = airbnb.groupby('room_type')['price']
grouped_price_summary = grouped_price_rt.describe()

grouped_price_summary['Median'] = grouped_price_rt.median()
grouped_price_summary['IQR'] = grouped_price_rt.apply(lambda x: x.quantile(0.75) - x.quantile(0.25))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

grouped_price_summary[['Median', 'IQR']].plot(kind='bar', ax=axes[0], color=['Blue', 'Orange'])
axes[0].set_title('Price by Room Type: Median and IQR')
axes[0].set_ylabel('Price')
axes[0].set_xlabel('Room Type')
axes[0].tick_params(axis='x', rotation=0)

sns.boxplot(data=airbnb, x='room_type', y='price', palette='pastel', showfliers=False, ax=axes[1])
axes[1].set_title('Price Distribution by Room Type')
axes[1].set_ylabel('Price')
axes[1].set_xlabel('Room Type')

plt.tight_layout()
plt.show()
````

## Borough Number of Reviews (Variance,Range,Mode)

````python
import matplotlib.pyplot as plt
import seaborn as sns

grouped_reviews = airbnb.groupby('borough')['number_of_reviews']
grouped_reviews_summary = grouped_reviews.describe()

grouped_reviews_summary['Variance'] = grouped_reviews.apply(lambda x: x.var())
grouped_reviews_summary['Range'] = grouped_reviews.apply(lambda x: x.max() - x.min())
grouped_reviews_summary['Mode'] = grouped_reviews.apply(lambda x: x.mode().iloc[0] if not x.mode().empty else None)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

grouped_reviews_summary[['mean', 'Mode']].plot(kind='bar', ax=axes[0], color=['Blue', 'Orange'])
axes[0].set_title('Reviews by Borough: Mean and Mode')
axes[0].set_ylabel('Number of Reviews')
axes[0].set_xlabel('Borough')
axes[0].tick_params(axis='x', rotation=45)

grouped_reviews_summary[['Variance', 'Range']].plot(kind='bar', ax=axes[1], color=['Green', 'Red'])
axes[1].set_title('Dispersion by Borough: Variance and Range')
axes[1].set_ylabel('Value')
axes[1].set_xlabel('Borough')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
sns.violinplot(data=airbnb, x='borough', y='number_of_reviews', palette='pastel')
plt.title('Number of Reviews Distribution by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Reviews')
plt.ylim(0, 200)
plt.show()
````
